(3-4) Let $a_1$ a positive integer. Consider a series:
$$
a_{n+1} = \begin{cases} a_n/2 & \text{if } a_n \text{ is even,} \\ 3a_n+1 &\text{if } a_n \text{ is odd.} \end{cases}
$$
The Collatz conjecture tells for every positive integer $a_1$, this series at some points comes to $1$, but it is unsolved till now (an computer experiment showed that the proposition is true for all integers less than as large as $2^{68}$). Write a program that, with an given integer input, returns the minimal number of the steps one has to take until the series gets to $1$. Try $a_1=7$ and $a_1=31$ to see how many steps to converge (these steps may take more than you would expect).


```python
a = 31
n = 0
for i in range(1000):
    if a == 1:
        break

    if a % 2 == 0:
        a = a // 2
    else:
        a = 3 * a + 1
    n = n + 1
print(n)
```

```python
a = 31
n = 0
while a != 1:
    if a % 2 == 0:
        a = a // 2
    else:
        a = 3 * a + 1
    n = n + 1
print(n)
```

(4-3)
Define a function `f` that takes a list as an argument `x` and returns a list of elements of `x` in reverse order. However, achieve this without using functions like `reversed`, but using loops or list slicing.

```python
x = [3, 9, 7, 1, 0]
```

```python
def f(x):
    y = []
    for i in range(len(x)):
        y.append(x[-i-1])
    return y

f(x)
```
